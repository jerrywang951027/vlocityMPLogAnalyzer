```python
# Display call tree with indentation
def print_tree(calls, max_depth=10, min_duration=0):
    """Print call tree with indentation"""
    def print_node(call, depth=0):
        if depth > max_depth or call.duration < min_duration:
            return
        
        indent = "  " * depth
        print(f"{indent}├─ [{call.duration:.0f}ms] {call.method_name.split('(')[0]}")
        
        for child in call.children:
            print_node(child, depth + 1)
    
    for root in calls:
        print_node(root)

# Display the tree
print("🌳 Call Tree:\n")
print_tree(call_tree, max_depth=5, min_duration=100)
```

```python
# Get critical path
critical_path = parser.get_critical_path()

# Display as tree (showing parent-child relationships)
print("🔥 Critical Path Tree:\n")

def print_critical_tree():
    """Print critical path with proper tree indentation"""
    for i, call in enumerate(critical_path):
        # Determine if this is the last sibling or has more
        is_last = (i == len(critical_path) - 1)
        connector = "└── " if is_last else "├── "
        indent = "│   " * call.level
        
        method_name = call.method_name.split('(')[0]
        print(f"{indent}{connector}[{call.duration:.0f}ms] {method_name}")

print_critical_tree()
```

```python
# Get critical path
critical_path = parser.get_critical_path()

print("🔥 Critical Path:\n")

# Track depth as we go through the list
prev_level = -1
depth = 0

for i, call in enumerate(critical_path):
    # Adjust depth based on level change
    if call.level > prev_level:
        depth += 1
    elif call.level < prev_level:
        depth = max(0, depth - 1)
    
    indent = "  " * depth
    connector = "└─ " if i == len(critical_path) - 1 else "├─ "
    method_name = call.method_name.split('(')[0]
    
    print(f"{indent}{connector}[{call.duration:.0f}ms] {method_name}")
    prev_level = call.level
```

```python
# Get critical path
critical_path = parser.get_critical_path()

print("🔥 Critical Path:\n")

# Build a map of call_id to position for proper nesting
call_positions = {call.call_id: i for i, call in enumerate(critical_path)}

# Track actual depth based on parent relationships
for i, call in enumerate(critical_path):
    # Find actual depth by looking at parent chain
    depth = 0
    current = call
    while current.parent_id and current.parent_id in call_positions:
        depth += 1
        # Find parent in critical path
        parent_idx = call_positions.get(current.parent_id)
        if parent_idx is not None and parent_idx < i:
            current = critical_path[parent_idx]
        else:
            break
    
    indent = "  " * min(depth, 10)
    method_name = call.method_name.split('(')[0]
    print(f"{indent}└─ [{call.duration:.0f}ms] {method_name}")
```

```python
critical_path = parser.get_critical_path()

print("🔥 Critical Path:\n")

# Build proper depth by checking actual containment
for i, call in enumerate(critical_path):
    method_name = call.method_name.split('(')[0]
    
    # Calculate depth by counting how many previous calls contain this one
    depth = 0
    for j in range(i):
        prev_call = critical_path[j]
        # Check if this call is temporally contained in prev_call
        if (prev_call.start_time <= call.start_time and 
            call.end_time <= prev_call.end_time):
            depth += 1
    
    indent = "  " * min(depth, 15)
    print(f"{indent}└─ [{call.duration:.0f}ms] {method_name}")
```
